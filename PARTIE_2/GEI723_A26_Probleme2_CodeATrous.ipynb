## Réseau de neurones Brian2

On peut maintenant créer un réseau Brian2. Commençons par définir quelques paramètres.

```python
# Fixons le seed aléatoire afin de pouvoir reproduire les résultats
# np.random.seed(0)
seed(0)

# Horloge de Brian2
defaultclock.dt = 0.5 * units.ms

# Cible de génération de code pour Brian2
# 'cython' nécessite un compilateur C++ (Visual Studio Build Tools sur Windows).
# 'numpy' fonctionne sans compilateur (plus lent).
# prefs.codegen.target = 'numpy'
prefs.codegen.target = 'cython'

time_per_sample =   0.35 * units.second
resting_time = 0.15 * units.second

v_rest_e = -65. * units.mV 
v_rest_i = -60. * units.mV 

v_reset_e = -65. * units.mV
v_reset_i = -45. * units.mV

v_thresh_e = -52. * units.mV
v_thresh_i = -40. * units.mV

refrac_e = 5. * units.ms
refrac_i = 2. * units.ms

tc_theta = 1e7 * units.ms
theta_plus_e = 0.05 * units.mV

tc_pre_ee = 20 * units.ms
tc_post_1_ee = 20 * units.ms
tc_post_2_ee = 40 * units.ms

# Taux d'apprentissage
nu_ee_pre =  0.0001 #to check maybe lower
nu_ee_post = 0.01 #to check

F = 0.25*Hz # Added F = 10 for later
N=100

number_of_epochs = 1
```

Définissons l'entrée au réseau de neurones. L'entrée est un encodeur de type codage par COMPLETER.

```python
period = time_per_sample + resting_time
train_rates = np.tile(X_train, (number_of_epochs, 1))
all_rates = np.vstack([train_rates, X_test]) * F
stimulus = TimedArray(all_rates, dt=period)

input_group = PoissonGroup(28*28, rates='stimulus(t, i) * int((t % period) < time_per_sample)') # Added func for poisson group
```

Définissons notre modèle de neurone ainsi que nos groupes de neurones.

```python
neuron_model = '''
    dv/dt = ((v_rest_e - v) + (I_synE + I_synI) / nS) / tau  : volt (unless refractory)

    I_synE =  ge * nS * -v           : amp
    
    I_synI =  gi * nS * (d_I_synI-v) : amp
    
    dge/dt = -ge/(1.0*ms)            : 1
    
    dgi/dt = -gi/(2.0*ms)            : 1
    
    tau                              : second (constant, shared)
    
    d_I_synI                         : volt (constant, shared)
    
    dtheta/dt = -theta / (tc_theta)  : volt

    w_sum                            : 1 #sum to speedup processing
'''

excitatory_group = NeuronGroup(
    N=N, model=neuron_model, refractory=refrac_e, 
    threshold='v >= v_thresh_e + theta', reset='v = v_reset_e; theta += theta_plus_e', method='euler') # Added threshold and reset for excitatory group and N
excitatory_group.tau = 100 * units.ms
excitatory_group.d_I_synI = -100. * units.mV

inhibitory_group = NeuronGroup(
    N=N, model=neuron_model, refractory=refrac_i, 
    threshold='v >= v_thresh_i', reset='v = v_reset_i', method='euler') # Added threshold and reset for inhibitory group and N
inhibitory_group.tau = 10 * units.ms
inhibitory_group.d_I_synI = -85. * units.mV
```

Et les synapses.

```python
synapse_model = "w : 1"

stdp_synapse_model = '''
    w : 1

    plastic : boolean (shared) # Activer/désactiver la plasticité
    
    post2before : 1
    
    dpre/dt    =   -pre / tc_pre_ee : 1 (event-driven)
    dpost1/dt  = -post1 / tc_post_1_ee : 1 (event-driven)
    dpost2/dt  = -post2 / tc_post_2_ee : 1 (event-driven)

    w_sum_post = w : 1 (summed)
'''

# dpre/dt   =   -pre / tau : 1 (event-driven)
    
    # dpost1/dt  = -post1 / tau : 1 (event-driven)
    
    # dpost2/dt  = -post2 / tau : 1 (event-driven)
    # COMPLETER

stdp_pre = '''
    ge_post += w
    
    pre = 1.
    
    w = clip(w - post1 * nu_ee_pre * plastic, 0, 1)
'''

stdp_post = '''
    post2before = post2
    
    w = clip(w + pre * nu_ee_post * post2before * plastic, 0, 1)
    
    post1 = 1.
    
    post2 = 1.
'''

input_synapse = Synapses(input_group, excitatory_group, model=stdp_synapse_model, on_pre=stdp_pre, on_post=stdp_post) # Added Synapses model
input_synapse.connect(True) # Fully connected
input_synapse.delay = 'rand()*5 * ms' # random delay or 10
input_synapse.plastic = True
input_synapse.w = 'rand()*0.3' #random weight maybe dc offset to weight
input_synapse.run_regularly('w = w * (28*28*0.1) / w_sum_post', dt=period, when='end') # Added run_regularly to normalize weights

e_i_synapse = Synapses(excitatory_group, inhibitory_group, model=synapse_model, on_pre='ge_post += w') # Added Synapses model
# e_i_synapse.connect(True, p=0.0025)
e_i_synapse.connect(j='i')
# e_i_synapse.w = 'rand()*10.4'
e_i_synapse.w = 10.4

i_e_synapse = Synapses(inhibitory_group, excitatory_group, model=synapse_model, on_pre='gi_post += w') # Added Synapses model
# i_e_synapse.connect(True, p=0.9)
i_e_synapse.connect(condition='i!=j')
# i_e_synapse.w = 'rand()*17.0'
i_e_synapse.w = 17.0
```

Définissons un 'readout' pour notre réseau.

```python
# e_monitor = SpikeMonitor(excitatory_group, record=False)
e_monitor = SpikeMonitor(excitatory_group)
```

Créons le réseau.

```python
net = Network(input_group, excitatory_group, inhibitory_group, 
              input_synapse, e_i_synapse, i_e_synapse, e_monitor)
```

## Entrainement

Entrainons à présent notre réseau.

Créons une matrice 'spikes' pour assigner les étiquettes des classes post-apprentissage. Cette matrice accumulera le décompte de décharges par classe.

Notre readout 'e_monitor' est actif tout le long de la simulation avec toutes les images. Donc à chaque nouvelle présentation d'image, on doit soustraire l'ancien décompte de décharges. On utilisera le vecteur 'old_spike_counts' pour le faire. 

```python
# Simuler tout le réseau (entraînement sur toutes les époques, puis test)
net.run(number_of_epochs * len(X_train) * period, report='text')
input_synapse.plastic = False   # Désactiver la plasticité STDP pour le test
net.run(len(X_test) * period, report='text')
device.build(compile=True, run=True)

# Décompte des décharges pour chaque échantillon présenté
n_presented = number_of_epochs * len(X_train) + len(X_test)
sample_idx = np.minimum((np.asarray(e_monitor.t / period)).astype(int), n_presented - 1)
counts = np.zeros((n_presented, len(excitatory_group)))
np.add.at(counts, (sample_idx, np.asarray(e_monitor.i)), 1)

spikes = np.zeros((10, len(excitatory_group)))

# Entrainement
for i in range(number_of_epochs):
    print('Starting iteration %i' % i)
    for j, (sample, label) in enumerate(zip(X_train, y_train)):
        # Enregistrer les décharges
        spikes[int(label)] += counts[i * len(X_train) + j]
```

```python
num_correct_output = 0
test_offset = number_of_epochs * len(X_train)   # les échantillons de test suivent toutes les époques

for i, (sample, label) in enumerate(zip(X_test, y_test)):
    # Calculer le nombre de décharges pour l'échantillon
    current_spike_count = counts[test_offset + i]

    # Prédire la classe de l'échantillon
    output_label = np.argmax(output_spike_count := current_spike_count[labeled_neurons])

    # Si la prédiction est correcte
    if output_label == int(label):
        num_correct_output += 1

print("The model accuracy is : %.3f" % (num_correct_output / len(X_test)))
```
